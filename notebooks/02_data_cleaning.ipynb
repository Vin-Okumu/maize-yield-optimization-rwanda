```python
import pandas as pd
import numpy as np

import warnings
warnings.filterwarnings("ignore")
```

We already know our raw datasets have some cleaning to be done, epsecially the date fields.

Now, let's clean and process the raw datasets and store them in the processed folder before merging.

```python
from pathlib import Path

# Defining path to folder
data_path = Path(r"C:\Users\admin\Documents\Projects\maize-yield-optimization-rwanda\data\raw")

def load_csv (filename):
    return pd.read_csv(data_path/filename)
```

```python
field_df = load_csv("Field_Level.csv")
field_df.head()
```

```python
display(field_df.info())
```

Since the field dataset has no data stored in the wrong format, and that the audit notebookshowed the key fields do not have a significant proportion of missing data, we can go ahead and store the cleaned staset in the interim folder awaiting merging.

```python
field_df.to_csv('../data/interim/field.csv', index = False)
```

```python
plot_df = load_csv("Plot_Level.csv")
date_cols = ['planting_date','germ_count_date','urea_V6_date','urea_V10_date',
             'silking_date','R3_date','harvest_date']

# Convert to datetime
plot_df[date_cols] = plot_df[date_cols].apply(pd.to_datetime, errors = 'coerce')

plot_df.info()
```
